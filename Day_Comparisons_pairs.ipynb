### Build a Corpus

Put all the files that interest you in the `Music_Files` folder.

Note that some are reporting errors when imported to Music21 and CRIM Intervals.  They are skipped.

````python
import glob
corpus_list = []
for name in glob.glob('Music_Files/*'):
    corpus_list.append(name)
corpus = CorpusBase(corpus_list)
corpus
````

### Get the Ngrams

Here we are including melodic intervals (diatonic) and durational ratios (which are the 'time intervals' between notes).

We can set the length of these ngrams to anything we like.  

And note that there we are limiting the ngrams to _entries only_.  That is:  melodic motives that begin after a rest.  We could certainly get _all_ ngrams, but these would be a vast number (and would nevertheless tell us something about style rather than identity).


### This is for Entry Ngrams (only those following a rest)

````python
# set length of desired ngrams
n = 10
func = ImportedPiece.durationalRatios
# get durational ratios
list_of_dur_rats = corpus.batch(func=func, metadata=True)
list_of_dur_rats_rounded = []
for df in list_of_dur_rats:
    list_of_dur_rats_rounded.append(df.round(2))
# get melodic intervals
func2 = ImportedPiece.melodic
list_of_mel = corpus.batch(func=func2, kwargs={'kind' : 'd'})
# make ngrams out of the melodic and durational ratios
func3 = ImportedPiece.ngrams
list_of_ngrs = corpus.batch(func=func3, kwargs={'df' : list_of_mel, 'other' : list_of_dur_rats_rounded, 'n' : n})
# get only the entries (things after rests)
func4 = ImportedPiece.entries
list_entries = corpus.batch(func=func4, kwargs={'df' : list_of_ngrs, 'thematic' : True, 'anywhere': True})
# combine them into one df
combined_df = pd.concat(list_entries, ignore_index=False)
combined_df
````

#### This One Gets ALL ngrams:  the "Moving Window"

````python
# ALL NGS
n = 10
func = ImportedPiece.durationalRatios
# get durational ratios
list_of_dur_rats = corpus.batch(func=func, metadata=True)
list_of_dur_rats_rounded = []
for df in list_of_dur_rats:
    list_of_dur_rats_rounded.append(df.round(2))
# get melodic intervals
func2 = ImportedPiece.melodic
list_of_mel = corpus.batch(func=func2, kwargs={'kind' : 'd'})
# make ngrams out of the melodic and durational ratios
func3 = ImportedPiece.ngrams
list_of_ngrs = corpus.batch(func=func3, kwargs={'df' : list_of_mel, 'other' : list_of_dur_rats_rounded, 'n' : n})
for ngrs in list_of_ngrs:
    ngrs.rename(columns={ngrs.columns[0]: 'ngram'}, inplace=True)


# combine them into one df
combined_df = pd.concat(list_of_ngrs, ignore_index=False)
combined_df
````

### Clean Up Columns and Rename Index

````python
combined_df = combined_df[['ngram', 'Composer', 'Title']]
combined_df.set_index(['Composer'], inplace=True)
len(combined_df)
combined_df
````

###  Grouping the Results by Ngrams

Each ngram now matched with a list of all the tunes that contain it

````python
grouped = combined_df.groupby(['ngram'])['Title'].apply(list).reset_index()
grouped
````

### Pairs from the Lists of Tunes that Share Ngrams

We take each list of pieces for a given tune and make pairs of each combination in that list.

So if Ngram Z is found in Tunes A, B, and C, we will have pairs A-B, A-C, and B-C.

We also take the _count_ of every unique pair of tunes.  Pairs that appear frequently are very likely to be related to each other.

````python
pairs = grouped.Title.apply(lambda x: list(combinations(x, 2)))
pairs2 = pairs.explode().dropna()
unique_pairs = pairs.explode().dropna().unique()
filtered_pairs = []
for pair in pairs2:
    if len(set(pair)) > 1:
          filtered_pairs.append(pair) 
df = pd.DataFrame(filtered_pairs, columns=["Tune_1", "Tune_2"])
pair_counts = df.value_counts().reset_index(name="Count")
````

## Grouping to Find % of Shared Ngrams for Every Pair of Melodies

````python
unique_values = combined_df.groupby(level=1)['ngram'].unique().dropna()

shared_values = {}
for idx1, vals1 in unique_values.iteritems():
    for idx2, vals2 in unique_values.iteritems():
        if idx1 != idx2:
            shared = len(set(vals1).intersection(vals2))
            total = len(set(vals1).union(vals2))
            shared_values[(idx1, idx2)] = shared / total * 100

result = pd.DataFrame.from_dict(shared_values, orient='index', columns=['Percentage'])
result.index = pd.MultiIndex.from_tuples(result.index, names=['Title1', 'Title2'])
result
````
