```python
file_list = glob.glob('Music_Files/*.mei')
```

```python
list_counts = []
for url in file_list:
    piece = importScore(url)
    
    non_na_count = piece.notes().count().sum()
    list_counts.append(non_na_count)
sum(list_counts)
```

```python
list_ngrams = []
for url in file_list:
    piece = importScore(url)
    mel = piece.melodic()
    count_mel_ngrams = piece.ngrams(df = mel, n=5).count().sum()
    list_ngrams.append(count_mel_ngrams)

sum(list_ngrams)
```

```python
list_dfs = []
for url in file_list:
    piece = importScore(url)
    mel = piece.melodic(kind='d')
    mel_ngrams = piece.ngrams(df = mel, n=5)
    list_dfs.append(mel_ngrams)

combined_df = pd.concat(list_dfs, axis=0, ignore_index=True)
list_of_tuples = combined_df.apply(tuple, axis=1).tolist()

# filtered_data = [tup for tup in list_of_tuples if not isinstance(tup, np.float64) or not np.isnan(tup)]
```
