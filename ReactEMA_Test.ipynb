```python
# function to build URL for each EMA ref

# NB must know the piece url!  could be passed in
def construct_url(row, piece_url):
    # Assuming 'item' and 'piece_url' are column names in your DataFrame
    mr = ''
    if mode == 'p_types':
        integers = re.findall(r'\d+', row['EMA'].split("/")[0])

        # Step 2: Extract the first and last elements and join them with a dash
        mr = f"{integers[0]}-{integers[-1]}"
    else:
        mr = row['EMA'].split("/")[0]
    pieceURL = piece_url
    ema_expression = "/" + row['EMA'] + "/highlight"
    measure_range = {"measureRange": mr}
    json_string = json.dumps(measure_range)
    encoded_mr = urllib.parse.quote(json_string)
    
    react_app_url = "https://eleon024.github.io/ema_react_app/"
    
    params = {
        "pieceURL": pieceURL,
        "ema_expression": ema_expression,
        "measure_range": encoded_mr
    }
    
    query_string = urllib.parse.urlencode(params)
    url = react_app_url + "?" + query_string
    return url
```

## Build EMA for A Given Function, add the full URL to react and make clickable

```python
# for melodic ngrams

piece_url = 'https://crimproject.org/mei/CRIM_Model_0008.mei'
piece = importScore(piece_url)
mode = 'melodic'
mel = piece.melodic()
ng = piece.ngrams(df=mel, n=4, offsets='both')
ema = piece.emaAddresses(df=ng, mode=mode)
ema_df = pd.DataFrame(ema)
result = ng.join(ema_df, how='inner')
# result.drop('EMA', axis=1, inplace=True)
result.fillna('', inplace=True)
ema_df['url'] = ema_df.apply(construct_url, piece_url = piece_url, axis=1)
# result.drop('EMA', axis=1, inplace=True)
result.fillna('', inplace=True)
# Define a function to format URLs as clickable links
def make_clickable(val):
    return '<a href="{}">Show Music</a>'.format(val, val)

# Apply the function to the URL column
result = result.rename(index=lambda val: round(val, 1))
result.index = result.index.droplevel(1)

result = result.head(5)

result.style.format({'url': make_clickable})
```

```python
# for cadences with react urls

piece_url = 'https://crimproject.org/mei/CRIM_Mass_0019_2.mei'
piece = importScore(piece_url)
cad = piece.cadences()
# ng = piece.ngrams(df=mel, n=4, offsets='both')
ema = piece.emaAddresses(df=ng, mode='cadence')
ema_df = pd.DataFrame(ema)
result = cad.join(ema_df, how='inner')
# result.drop('EMA', axis=1, inplace=True)
result.fillna('', inplace=True)
result['url'] = result.apply(construct_url, piece_url = piece_url, axis=1)
result.drop('EMA', axis=1, inplace=True)
result.fillna('', inplace=True)
# Define a function to format URLs as clickable links
def make_clickable(val):
    return '<a href="{}">Show Music</a>'.format(val, val)

# Apply the function to the URL column
result = result.rename(index=lambda val: round(val, 1))
result.index = result.index.droplevel(1)

result.style.format({'url': make_clickable})
```
