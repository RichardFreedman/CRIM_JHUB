## B. Imposta la ricerca

```python
# ESEGUI QUESTA CELLA
piece_list = ['CRIM_Model_0019',
                     'CRIM_Mass_0019_1',
                     'CRIM_Mass_0019_2',
                     'CRIM_Mass_0019_3',
                     'CRIM_Mass_0019_4',
                     'CRIM_Mass_0019_5']
print('Cella eseguita')
```

```python
# segui le istruzioni ed ESEGUI LA CELLA

# IMPOSTA:

# la lunghezza dei motivi (n)
n=4 # il numero deve corrispondere alla lunghezza degli n_grams che desideri cercare

# se attivare o disattivare il parametro combineUnisons
combineUnisons=True # usa True o False senza virgolette

# la tipologia degli intervalli (kind), se diationic ('d') o quality ('q'). Ricorda di usare le virgolette
kind='d'

# quali occorrenze selezionare:
# - se anywhere (dovunque) è impostato su True verranno incluse tutte le occorrenze di un motivo,
# - se è impostato su False verranno selezionate soltanto le occorrenze in corrispondenza dell'entrata di una voce, cioè dopo una pausa
anywhere=False

print('Cella eseguita')
```

```python
# segui le istruzioni ed ESEGUI LA CELLA

# COMPONI LA LISTA DEI SOGGETTI
# la lista si apre e si chiude con parentesi quadrate
# i soggetti devono essere scritti fra parentesi () e sono separati con virgola e spazio (oppure virgola e a capo)
# gli intervalli devono essere scritti fra virgolette '' e devono essere separati con virgola e spazio,
# assicurati che i soggetti siano tutti della stessa lunghezza, cioè che abbiano lo stesso numero di intervalli
# la lunghezza dei soggetti deve corrispondere al valore n impostato nella cella precedente

soggetti = [('-3', '3', '2', '-2'),
            ('-3', '2', '2', '-2'),
            ('-3', '3', '-2', '-2'),
            ('-2', '2', '3', '-2')]

print('Cella eseguita')
```

## C. Esegui la ricerca

La prossima cella esegue le seguenti operazioni:<br>
- trova all'interno del corpus (mottetto+messa) tutti gli n-gram della lunghezza e della tipologia impostata;<br>
- calcola la lunghezza di ciascun n-gram (rappresentata dallo spessore dei mattoncini nella mappa);<br>
- fra gli n-gram rilevati, seleziona soltanto quelli inclusi nella lista preparata in una delle celle precedenti;<br>
- in base al parametro anywhere (True o False) seleziona tutte le occorrenze o solo quelle dopo una pausa.<br>
<br>
Sulla base delle impostazioni, viene visualizzata una mappa in cui l'asse orizzontale rappresenta lo svolgimento del brano, l'asse verticale rappresenta l'ordine delle voci, i mattonicini rappresentano le occorrenze dei soggetti della lista (un colore per ogni soggetto) e la loro ampiezza rappresenta la durata di ciascuna occorrenza (l'unità di misura è la semiminima).<br>
Segui le istruzioni all'interno della cella per includere anche una traccia audio.

```python
# segui le istruzioni ed ESEGUI LA CELLA
# accanto a ogni grafico, in alto a destra, c'è un bottone che consente di scaricare l'immagine in vari formati (meglio SVG)
# clicca a sinistra dei grafici per attivare o disattivare lo scorrimento verticale
for piece in piece_list:
    prefix = 'https://crimproject.org/mei/' 
    mei_file = piece
    url = prefix + mei_file + '.mei'
    piece = importScore(url)
    nr = piece.notes(combineUnisons=True)
    mel = piece.melodic(df=nr, kind='d', compound=False, unit=0, end=False)
    mel_ngrams = piece.ngrams(df=mel, n=4)
    piece_entries = piece.entries(df=mel, n=4, anywhere=anywhere)
    mel_ngrams_duration = piece.durations(df=mel, n=4, mask_df=piece_entries)
    print(piece.metadata)
    display(viz.plot_ngrams_heatmap(piece_entries, mel_ngrams_duration,
                            selected_patterns=soggetti, voices=[]))
    #display(IPython.display.Audio('https://crimproject.org/static/mp3/' + mei_file + '.mp3'))
    # aggiungi o rimuovi il segno '#' all'inizio della riga precedente per escludere o includere la traccia audio
```
