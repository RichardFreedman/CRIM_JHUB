## Converts MEI 4 to 5

### Load Libraries

```python


from pathlib import Path
from saxonche import PySaxonProcessor
import glob as glob
import os
```

### Set Source Folders and XSLT

```python

xsl_path = 'mei40To50.xsl'
output_folder = 'Output'
```

### The Functions

```python



def modify_path(path):
    # Split the path into directory and filename
    directory, filename = os.path.split(path)
    
    # Split the filename into name and extension
    name, extension = os.path.splitext(filename)
    
    # Insert "_mei5" before the "." and extension
    new_filename = f"{name}_mei5{extension}"
    
    # Join the directory and the new filename to form the new path
    # new_path = os.path.join(directory, new_filename)
    
    return new_filename


def transform_xml_with_xslt3(xml_path, xsl_path, output_folder):
    # Ensure the output directory exists
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # get updated filename:
    output_file_name = modify_path(xml_path)
    
    # Construct the full path to the output file
    output_path = os.path.join(output_folder, output_file_name)
    
    with PySaxonProcessor(license=False) as proc:
        xsltproc = proc.new_xslt30_processor()
        document = proc.parse_xml(xml_text=open(xml_path, 'r').read())
        executable = xsltproc.compile_stylesheet(stylesheet_file=xsl_path)
        output = executable.transform_to_string(xdm_node=document)
        with open(output_path, 'w') as output_file:
            output_file.write(output)
```

### Now Load the Files and Run Function

```python

corpus_list = sorted(glob.glob('Music_Files/*.mei'))
for piece in corpus_list:
    transform_xml_with_xslt3(piece, xsl_path, output_folder)
```
