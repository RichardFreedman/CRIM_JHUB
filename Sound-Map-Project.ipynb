```python
#Installing libraries
# !pip install folium
import folium
import folium.plugins as plugins
from folium import IFrame
import pandas as pd
#look up on pypi for the library in question to find how to install the library in question
```

```python
#Importing csv file from github
url = "https://raw.githubusercontent.com/jojoag10/encoding-music-project/main/Encoding%20Music_%20Sound%20Map%20of%20Haverford%203%20-%20Encoding%20Music_%20Sound%20Map%20of%20Haverford%203%20(2).csv"
dataset = pd.read_csv(url)
```

```python
dataset.columns = ["Submitted","Link","Time","Date","Location","Longitude","Latitude","Device","Sound","Recorder","Purpose","Volume","Distractability","Punctuability","Tempo","Rowdiness","Incidentality","Classification"]
dataset
```

```python
def makeMessage(df, indx):

    
    message = "Link to sound: <br><br>" + dataset["Link"][indx] # Links sound file
    message += "<br><br>Recording information: <br><br> Description: <br>>" + dataset["Sound"][indx] # Provides description of the sound
    message += "<br> Time Recorded: <br>>" + dataset["Time"][indx] # Provides time of day it was recorded
    message += "<br> Date Recorded: <br>>" + dataset["Date"][indx] # Provides date it was recorded
    message += "<br> Recorded by: <br>>" + dataset["Recorder"][indx] # Provides recorder
    message += "<br> Recorded on: <br>>" + dataset["Device"][indx] # Provides recorded device
    message += "<br><br> Stats: <br><br> Original sound emitted for: "+ dataset["Purpose"][indx] #Purpose
    message += "<br> Volume, from 1-10: " + str(dataset["Volume"][indx])
    message += "<br> Distractability, from 1-10: " + str(dataset["Distractability"][indx])
    message += "<br> Punctuability, from 1-10: " + str(dataset["Punctuability"][indx])
    message += "<br> Tempo, from 1-10: " + str(dataset["Tempo"][indx])
    message += "<br> Rowdiness, from 1-10: " + str(dataset["Rowdiness"][indx])
    message += "<br> Is the sound incidental? " + dataset["Incidentality"][indx]
    message += "<br> Classification: " + dataset["Classification"][indx]
    return message

```

Below is the entire list of collected datapoints; can click, zoom in and zoom out, and look at everything individually.

```python
#Used this post for help with tooltip formatting https://stackoverflow.com/questions/65524514/how-can-we-get-tooltips-and-popups-to-show-in-folium
m = folium.Map(location=[40.009395, -75.305613],zoom_start=17)
tooltip = "Click me!"
for indx in dataset.index:
  lon = dataset["Longitude"][indx]
  lat = dataset["Latitude"][indx]
  message = makeMessage(dataset, indx)
  folium.Marker([lon,lat], popup=message, tooltip=tooltip).add_to(m)
m
```

Here we have our first visualization of data; this map consists of every recorsding taken for the project. It's clear that we got a nice spread of campus; while a perfect project would have taken many more samples over many more days, this remains a nice data spread for the limittions we were working under! Each entry can be clicked on to give details about our ratings, scalars, and decriptions of the sound, as well as a link to the recording itself on Google Drive. Loudness

```python
#Color codes in intervals of 2
def colorCode(value, indx):
  if (dataset[value][indx]  <=  2):
    color='white'
  elif (dataset[value][indx]  <=  4):
    color='lightblue'
  elif (dataset[value][indx]  <=  6):
    color='green'
  elif (dataset[value][indx]  <=  8):
    color='orange'
  else:
    color ='red'
  return color

import datetime

def colorCodeTime(indx):
  time = datetime.datetime.strptime(dataset["Time"][indx],"%H:%M")
  if (time  <=  datetime.datetime.strptime("6:00", "%H:%M")):
    color='red'
  elif (time  <=  datetime.datetime.strptime("11:00", "%H:%M")):
    color='white'
  elif (time  <=  datetime.datetime.strptime("15:00", "%H:%M")):
    color='lightblue'
  elif (time  <=  datetime.datetime.strptime("18:00", "%H:%M")):
    color='green'
  elif (time  <=  datetime.datetime.strptime("21:00", "%H:%M")):
    color='orange'
  else:
    color ='red'
  return color
```

```python
# Color sorted by loudness, white -> light blue -> green -> orange -> red in that order
m = folium.Map(location=[40.009395, -75.305613],zoom_start=17)
for indx in dataset.index:
  lon = dataset["Longitude"][indx]
  lat = dataset["Latitude"][indx]
  message = "Volume: " + str(dataset["Volume"][indx])  
  message += "<br><br>Description: <br>>" + dataset["Sound"][indx] # Provides description of the sound
  message += "<br><br>Link to sound: <br>" + dataset["Link"][indx] # Links sound file
  
  folium.Marker([lon,lat], popup=message, tooltip=tooltip,icon=folium.Icon(color=colorCode("Volume",indx),icon_color='#FFFF00')).add_to(m)
m
```

Our next project was splicing the data, and the folliwng maps display a simple binning and sorting color scheme that plays on the scalars we assigned all of our data points. White corresponds to a low value on whatever scale we're measuring, while red is the highest possible value. This map displays volume, and lends itself to some interesting conclusions. While it is important to acknowledge that neither of us are objective observers, we tried to be as systematic as possible in assigning values to these sounds. This map suggests that campus has a wide spread of sound, but that a listener is more likely to reach extremes as they move to the edges of campus. While some areas on the edge of Haverford College, like the Ryan Pinetum, are relativley quiet, other areas—like the apartments on a weekend—are quite loud. The center of campus also has a few loud points, particularly near dorms (again, we can thank partying for this). Noises from inside of academic buildings tend to be quieter, which makes sense; it is rare indeed that a professor lectures as loudly as their students party on the weekend. This map thus paints a picture of Haverford as a dynamic place, where students get up to loud mischief and professors quietly teach within yards of each other, depending on when one observes campus. This dynamic is fairly unique to the college campus space, and could be an entryway for additional analysis; a shifting, chimeric space, Haverford has many sonic sides to be explored.

```python
# Color sorted by time of day. white is morning (6-11)-> light blue is afternoon (11-15) -> green is late afternoon (15-18) -> orange is evening (18-211) -> red is night (21-6)
# Split into two maps: The left map shows all quieter sounds (volume rating less than 5), while the right map shows all louder sounds (volume rating of 5 or greater).
m = folium.plugins.DualMap(location=[40.009395, -75.305613],zoom_start=17)
for indx in dataset.index:
  lon = dataset["Longitude"][indx]
  lat = dataset["Latitude"][indx]
  message = "Volume: " + str(dataset["Volume"][indx])  
  message += "<br>Time: " + dataset["Time"][indx]
  message += "<br><br>Description: <br>>" + dataset["Sound"][indx] # Provides description of the sound
  message += "<br><br>Link to sound: <br>" + dataset["Link"][indx] # Links sound file
  if dataset["Volume"][indx] < 5:
    folium.Marker([lon,lat], popup=message, tooltip=tooltip,icon=folium.Icon(color=colorCodeTime(indx),icon_color='#FFFF00')).add_to(m.m1)
  else:
    folium.Marker([lon,lat], popup=message, tooltip=tooltip,icon=folium.Icon(color=colorCodeTime(indx),icon_color='#FFFF00')).add_to(m.m2)
m
```

This map simply divides the data up into times of day, with the right version displaying louder sounds, and the left displaying quieter ones. Reviewing the data we've collected, this analyst would tentativley suggest that morning may be the most peaceful time on campus, as most students are asleep. As campus wakes up, it moves into a bustling academic midday, then slackens off again into the afternoon. Only with the arrival of evening does campus flare to life again, particularly if it is a Friday or Saturday; this pattern is attested to by the overwhelming presence of red dots on the rightmost map, which correspond to later times in the day.
